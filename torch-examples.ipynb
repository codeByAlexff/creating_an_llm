```python
import torch
import numpy as np
from torch import nn
from torch.nn import functional as F
import time
```

```python
#Dot Product / Matrix Multiplication
a = torch.tensor([[1,2],[3,4],[5,6]])
b = torch.tensor([[7,8,9],[10,11,12]])
print(a @ b)
print(torch.matmul(a,b))
```

```python
#Floats and Ints cannot mix unless you cast float onto int.
#Matrix multiplications will be done with floating point numbers and not ints
#Type int64
int_64 = torch.randint(1,(3,2)).float()

#Type float32
float_32 = torch.rand(2,3)
print(int_64.dtype, float_32.dtype)
result = torch.matmul(int_64, float_32)
print(result)
```

```python
x = torch.tensor([-0.05], dtype=torch.float32)

n = nn.ReLU()

#What RELU does:
#If a number is 0 or below 0 it will turn the num into 0
#If it's above 0 it will stay the same
#It is a non-linear activation function

#ReLU is Rectified Linear Unit - Direct Shape, not S curve
#Range: [0,infinity]
y = n(x)
print(y)
```

```python
#Sigmoid Non-Linear Activation Function
#Turns any number into a value between 0 and 1

#Sigmoid has an S-shape curve
#Range: [0,1]
y = F.sigmoid(x)
print(y)
```

```python
#Tanh Non-Linear Activation Function
#Turns any real value number into a range between -1 and 1

#Tanh has an almost S-shape curve like Sigmoid but different ranges
#Range: [-1,1]
y = F.tanh(x)
print(y)
```
